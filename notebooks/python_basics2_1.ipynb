```python
closing_prices=[2100,2150,2080,2200,2175,2250,2190]

first_price=closing_prices[0]
last_price=closing_prices[-1]

recent_3=closing_prices[-3:]
first_3=closing_prices[:3]

print(f"共{len(closing_prices)}天数据")

closing_prices.append(2210)

average=sum(closing_prices)/len(closing_prices)
print(f"平均:{average}")

daily_changes=[closing_prices[i]-closing_prices[i-1] for i in range(1,len(closing_prices))]
print(f"每日涨跌:{daily_changes}")
up_days=[n+2 for n in range(0,len(daily_changes)) if daily_changes[n] > 0 ]
print(f"涨日统计:{up_days}")
```

```python
prices = [2100, 2150, 2080, 2200, 2175, 2250, 2190, 2210, 2180, 2230]

sma5=[]
for i in range(4,len(prices)):
    window=prices[i-4:i+1]
    avg=sum(window)/5
    sma5.append(avg)
print(f"sma5:{sma5}")
```
