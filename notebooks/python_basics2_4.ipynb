```python
import pandas as pd
import numpy as np
import sys
sys.path.append(r'D:\start_up\git_practice\quant_trading_lab')

from src.data.fetcher import fetch_crypto_data,save_data,load_data
eth = pd.read_csv(r"D:\start_up\git_practice\quant_trading_lab\data\raw\eth_daily.csv", index_col=0,parse_dates=True)
eth.index.name = "date"

print("数据形状:",eth.shape)
print("\n前五行:")
eth.head()
print(eth.columns.tolist())

print(eth.describe())
print(f"共{len(eth)}个交易日")

close=eth['close']
print(f"最新价格:{close.iloc[-1]:.2f}")
print(f"平均价格:{close.mean():.2f}")
print(f"价格标准差:{close.std():.2f}")

print(f"\n平均日涨跌幅:{close.pct_change().mean():.4f}%")
print(f"涨跌幅标准差:{close.pct_change().std():.2f}")
```

```python
eth['daily_return']=eth['close'].pct_change()*100
eth['is_up']=eth['close']>eth['open']
eth['body_size']=abs(eth['close']-eth['open'])
eth['upper_shadow']=eth['high']-eth[['open','close']].max(axis=1)
eth['lower_shadow']=eth[['open','close']].min(axis=1)-eth['low']

eth[['close','daily_return','is_up','body_size','upper_shadow','lower_shadow']].tail(5)
eth_2025=eth.loc['2025']
print(f'2025年共{len(eth_2025)}个交易日')
eth_recent=eth.loc['2026-04-01':'2026-05-20']
print(f'指定范围:{len(eth_recent)}个交易日')

avg_volume=eth["volume"].mean()
bullish_volume=eth[(eth['daily_return']>2)&(eth['volume']>avg_volume*1.5)]
print(f'放量上涨:{len(bullish_volume)}日')
```

```python
eth_2025 = eth.loc['2025']
monthly_avg = eth_2025['close'].resample('ME').mean()
print(monthly_avg)

eth['daily_return']=eth['close'].pct_change()*100
by_weekday=eth['daily_return'].groupby(eth.index.dayofweek).mean()
by_weekday.index=['Mon','Tue','Wed','Thur','Fri','Sat','Sun']
print(f"\n涨幅最大:{by_weekday.idxmax()}{by_weekday.max():.4f}%")
print(f"\n涨幅最小:{by_weekday.idxmin()}{by_weekday.min():.4f}%")

top10_volume = eth.nlargest(10, 'volume')
```
