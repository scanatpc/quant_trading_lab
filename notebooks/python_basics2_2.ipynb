```python
import random
prices=[2100,2150,2080,2200,2175,2250,2190]
###
all_time_high=0
new_highs=[]

for i,price in enumerate(prices):
    if price>all_time_high:
        all_time_high=price
        new_highs.append({"DAY":i+1,"PRICE":price})
print(new_highs)
###
for i in range(1,len(prices)):
    change=prices[i]-prices[i-1]
    pct=change/prices[i-1]*100
    direction="↑↑" if change>0 else "↓↓"
    print(f"DAY{i+1}:PRICES:{prices[i]}({direction}{abs(pct):.1f}%)")

entry_price=2200
stop_loss=2150
random.seed(42)
print(f"\n入场:{entry_price} 止损:{stop_loss}")
###
current=entry_price
day=0
history=[]

while current>stop_loss:
    day+=1
    change=random.uniform(-50,50)
    current=current+change
    history.append({'DAY':day,'PRICE':round(current,2)})
    
print(history)    
print(f"第{day+1}天触发止损 亏损{current-entry_price:.2f}")    
    
```
